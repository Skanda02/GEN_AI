Train a custom Word2Vec model on a small dataset. Train embeddings on a
domain-specific corpus (e.g., legal, medical) and analyze how embeddings capture domain-
specific semantics.

```python
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess

# Small domain-specific corpus
sentences = [
    "legal contract agreement court law",
    "court judge legal case law",
    "medical doctor patient hospital treatment",
    "medical report patient diagnosis doctor"
]

# Tokenize
data = [simple_preprocess(s) for s in sentences]

# Train model
model = Word2Vec(data, vector_size=50, min_count=1)

# Find similar words
print(model.wv.most_similar("legal", topn=5))
```
