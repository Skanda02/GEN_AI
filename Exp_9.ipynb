Take the Institution name as input. Use Pydantic to define the schema for the
desired output and create a custom output parser. Invoke the Chain and Fetch Results.
Extract the below Institution related details from Wikipedia: The founder of the Institution.
When it was founded. The current branches in the institution. How many employees are
working in it. A brief 4-line summary of the institution.

```python
from pydantic import BaseModel
import wikipediaapi

# Define schema
class InstitutionDetails(BaseModel):
    founder: str | None
    founded: str | None
    summary: str | None

# Fetch data
def get_details(name):
    wiki = wikipediaapi.Wikipedia(
     user_agent="SkandaNLPProject/1.0 (skanda@email.com)",
     language="en"
    )    
    page = wiki.page(name)

    if not page.exists():
        print("Page not found")
        return None

    text = page.text.lower()

    # Simple extraction (basic but works for lab)
    founder = "founder" if "founder" in text else None
    founded = "founded" if "founded" in text else None

    return InstitutionDetails(
        founder=founder,
        founded=founded,
        summary=page.summary[:300]
    )

# Main
name = input("Enter institution: ")
data = get_details(name)

if data:
    print("\nDetails:\n", data)
```
