Use word embeddings to improve prompts for Generative AI model. Retrieve
similar words using word embeddings. Use the similar words to enrich a GenAI prompt. Use
the AI model to generate responses for the original and enriched prompts. Compare the
outputs in terms of detail and relevance.

```python
import gensim.downloader as api
from transformers import pipeline

# Load models
word_vectors = api.load("glove-wiki-gigaword-50")
generator = pipeline("text-generation", model="gpt2")

# Original prompt
prompt = "Who is king"

# Get similar word
similar_word = word_vectors.most_similar("king", topn=1)[0][0]

# Create enriched prompt
enriched_prompt = prompt.replace("king", similar_word)

print("Original Prompt:", prompt)
print("Enriched Prompt:", enriched_prompt)

# Generate responses
original_output = generator(prompt, max_length=50)[0]['generated_text']
enriched_output = generator(enriched_prompt, max_length=50)[0]['generated_text']

# Print outputs
print("\nOriginal Response:\n", original_output)
print("\nEnriched Response:\n", enriched_output)

# Simple comparison
print("\nLength Comparison:")
print(len(original_output), "vs", len(enriched_output))
```
