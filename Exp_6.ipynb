Use a pre-trained Hugging Face model to analyze sentiment in text. Assume a real-
world application, Load the sentiment analysis pipeline. Analyze the sentiment by giving
sentences to input.

```python
from transformers import pipeline

# Load sentiment model
analyzer = pipeline("sentiment-analysis")

# Sample inputs (real-world: customer reviews)
reviews = [
    "The product is amazing! I love it.",
    "I'm very disappointed. The service was terrible.",
    "It was an average experience.",
    "Absolutely fantastic quality!",
    "Not great, but okay."
]

# Analyze sentiment
for r in reviews:
    result = analyzer(r)[0]
    print(f"{r} → {result['label']} ({result['score']:.2f})")
```
